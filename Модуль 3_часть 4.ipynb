# Level 1

```python
import json
import os

def register(login, passwd):
    # Структура данных: словарь {логин: пароль}
    filename = "users.json"
    
    # Загружаем существующих пользователей
    if os.path.exists(filename):
        with open(filename, 'r', encoding='utf-8') as f:
            try:
                users = json.load(f)
            except json.JSONDecodeError:
                users = {}
    else:
        users = {}
    
    # Добавляем нового пользователя
    users[login] = passwd
    
    # Сохраняем в файл
    with open(filename, 'w', encoding='utf-8') as f:
        json.dump(users, f, ensure_ascii=False, indent=2)
    
    print(f"Пользователь '{login}' успешно зарегистрирован!")
```

# Level 2

```python
import json
import os

def register(login, passwd):
    filename = "users.json"
    
    if os.path.exists(filename):
        with open(filename, 'r', encoding='utf-8') as f:
            try:
                users = json.load(f)
            except json.JSONDecodeError:
                users = {}
    else:
        users = {}
    
    if login in users:
        print(f"Ошибка: Пользователь с логином '{login}' уже существует!")
        return False
    
    users[login] = passwd
    
    with open(filename, 'w', encoding='utf-8') as f:
        json.dump(users, f, ensure_ascii=False, indent=2)
    
    print(f"Пользователь '{login}' успешно зарегистрирован!")
    return True
```

# Level 3

```python
import json
import os

def login_function(login, passwd):
    filename = "users.json"
    
    if not os.path.exists(filename):
        print("Файл с пользователями не найден!")
        return False
    
    with open(filename, 'r', encoding='utf-8') as f:
        try:
            users = json.load(f)
        except json.JSONDecodeError:
            print("Ошибка чтения файла пользователей!")
            return False
    
    if login in users and users[login] == passwd:
        print(f"Вход выполнен успешно! Добро пожаловать, {login}!")
        return True
    else:
        print("Ошибка: Неправильный логин или пароль!")
        return False
```
