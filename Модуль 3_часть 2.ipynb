# Level 1

```python
I = [1, 4, 1, 6, "hello", "a", 5, "hello"]

duplicates = []

for i in range(len(I)):
    if I[i] in I[i+1:] and I[i] not in duplicates:
        duplicates.append(I[i])

print(duplicates)
```

# Level 2

```python
from random import randint

n = 5
m = [[randint(0, 100) for i in range(n)] for j in range(n)]

print("Матрица:")
for row in m:
    print(row)

max_element = m[0][0]

for row in m:
    for element in row:
        if element > max_element:
            max_element = element

print(max_element)
```

# Level 3

```python
d = {'name1': 'id1', 'name2': 'id2', 'name3': 'id3'}

reversed_dict = {}

for key, value in d.items():
    reversed_dict[value] = key

print(d)
print(reversed_dict)
```
