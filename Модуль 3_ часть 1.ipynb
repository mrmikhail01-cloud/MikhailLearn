# Level 1

```python
x = int(input("Введите начальную сумму вклада (рубли): "))
y = int(input("Введите целевую сумму (рубли): "))
p = float(input("Введите процент: "))

years = 0
while x < y:
    x = int(x * (1 + p / 100))
    years += 1

print(years)
```

# Level 2

```python
n = int(input("Введите количество повторений: "))
count = 0

while count < n:
    print('for - частный случай цикла while')
    count += 1
```

# Level 3

```python
number = input("Введите целое число: ")
sum_digits = 0

for digit in number:
    if digit.isdigit():  # проверяем, что символ - цифра (на случай знака минус)
        sum_digits += int(digit)

print(sum_digits)
```
