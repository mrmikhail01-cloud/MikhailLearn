```python
import pandas as pd
from pathlib import Path

# Загрузка данных
BASE_DIR = Path(r"C:\Users\mrmik\Downloads\titanic")
df = pd.read_csv(BASE_DIR / "train.csv")
# Для расчета берем только тех, у кого известен возраст
df_age = df.dropna(subset=['Age'])
# Задаем границы групп: 0-25, 25-50, 50+
bins = [0, 25, 50, 100]
labels = ['0-25 лет', '25-50 лет', '50+ лет']
df_age['AgeGroup'] = pd.cut(df_age['Age'], bins=bins, labels=labels, right=False)
# Считаем вероятности
probabilities = df_age.groupby('AgeGroup', observed=True)['Survived'].mean()
print("Уровень Easy: Вероятность выживания по возрастам")
for group, prob in probabilities.items():
    print(f"{group}: {prob:.2%}")
```

```python
print("\n Уровень Normal: Теорема Байеса")

# Считаем нужные вероятности на основе всех данных
p_female = len(df[df['Sex'] == 'female']) / len(df)
p_survived_given_female = df[df['Sex'] == 'female']['Survived'].mean()
p_survived = df['Survived'].mean()
# Формула Байеса
p_female_given_survived = (p_survived_given_female * p_female) / p_survived
print(f"Вероятность, что выживший пассажир — женщина: {p_female_given_survived:.2%}")
# (Сходится с логикой: женщин выжило 233, всего выживших 342. 233/342 = 68.1%)
```

```python
import numpy as np
from scipy.integrate import quad

print("\n Уровень Hard: Интегрирование")
# Предположим, что параметр lambda = 1
lambda_param = 1.0
# Задаем функцию плотности
def f(x):
    if x > 0:
        return lambda_param * np.exp(-lambda_param * x)
    return 0
# Вычисляем интеграл от 1 до 2
result, error = quad(f, 1, 2)
print(f"Параметр lambda = {lambda_param}")
print(f"Вероятность попадания в интервал [1, 2]: {result:.6f}")
print(f"Аналитическое решение: e^(-{lambda_param}) - e^(-{2*lambda_param})")
```
