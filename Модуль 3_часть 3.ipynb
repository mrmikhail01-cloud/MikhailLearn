# Level 1

```python
import math

def area(a, b, c):
    if a + b > c and a + c > b and b + c > a:
        p = (a + b + c) / 2
        s = math.sqrt(p * (p - a) * (p - b) * (p - c))
        return s
    else:
        return "Треугольник с такими сторонами не существует"

a = float(input("Введите первую сторону: "))
b = float(input("Введите вторую сторону: "))
c = float(input("Введите третью сторону: "))

result = area(a, b, c)
print(f"Площадь треугольника: {result}")
```

# Level 2

```python
def short_words(text):
    for char in '.,!?-...\n':
        text = text.replace(char, ' ')
    
    words = text.split()
    
    result = []
    for word in words:
        if len(word) < 5:
            result.append(word)
    
    return result

s = '''Было просто пасмурно, дуло с севера
А к обеду насчитал сто градаций серого.
Так всегда первого ноль девятого
То ли весь мир сошёл с ума, то ли я - того...
На столе записка от неё смятая
Недопитый виски допиваю с матами.
Посмотрю в окно, допишу главу
Первое сентября, дворник жжёт листву.
Серым облакам наплевать на нас
Если знаешь как жить - живи
Ты хотела плыть как все - так плыви!'''

short = short_words(s)
print(short)
```

# Level 3

```python
def make_max_number(numbers):
    str_nums = [str(num) for num in numbers]
    str_nums.sort(key=lambda x: x*10, reverse=True)
    
    result = ''.join(str_nums)
    
    return result

numbers1 = [56, 9, 11, 2]
numbers2 = [3, 81, 5]

print(make_max_number(numbers1))
print(make_max_number(numbers2))
```
